```
use role data_engineer;
use warehouse xs_wh;
use schema ACME.RAW;
SELECT * FROM ACME.RAW.DAILY_IMPRESSIONS LIMIT 100; 
```

```
from snowflake.snowpark.context import get_active_session
session = get_active_session()
df = session.table("ACME.RAW.DAILY_IMPRESSIONS")
df = df.with_column("COST_PER_CLICK", (df["SPEND"] / df["CLICKS"]).cast("DECIMAL(10,2)"))
df.select("DATE", "CAMPAIGN_ID", "CHANNEL", "CLICKS", "SPEND", "COST_PER_CLICK").sort("DATE").show(20)
```
